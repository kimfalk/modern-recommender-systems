# Section 3.3.1 – How Matrix Factorization Works

This notebook walks through the small example from Section 3.3.1.
We manually define user and item factor vectors, compute dot products,
and show how matrix factorization predicts missing ratings.

```python
import numpy as np
```

## User and item factors (k=2)

These are the factor vectors from Table 3.1 in the book.
Peter and Phylla like action films; Mela prefers family-friendly movies.

```python
# User factors
Peter  = np.array([0.9, -0.1])
Phylla = np.array([0.8, -0.2])
Mela   = np.array([-0.1, 0.9])

# Item factors
pulp_fiction = np.array([0.9, -0.9])
fight_club   = np.array([0.8, -0.2])
toy_story    = np.array([-0.1, 0.9])
notebook     = np.array([-0.1, 0.8])
```

## Predicting ratings with dot products

The dot product of a user vector and an item vector approximates the rating.

```python
print("Peter's predicted ratings:")
print(f"  Peter × Pulp Fiction = {np.dot(Peter, pulp_fiction):.2f}")
print(f"  Peter × Fight Club   = {np.dot(Peter, fight_club):.2f}")
print(f"  Peter × Toy Story    = {np.dot(Peter, toy_story):.2f}")
print(f"  Peter × Notebook     = {np.dot(Peter, notebook):.2f}  ← prediction for missing rating")
```

```python
# Verify the ordering matches what we expect
assert np.dot(Peter, pulp_fiction) > np.dot(Peter, fight_club) > np.dot(Peter, toy_story)
print("✓ Peter prefers Pulp Fiction > Fight Club > Toy Story (as expected)")
```

## All users × all items

Let's compute the full predicted rating matrix by multiplying
the user factor matrix by the item factor matrix.

```python
import pandas as pd

users = np.array([Peter, Phylla, Mela])
items = np.array([pulp_fiction, fight_club, toy_story, notebook])

predicted_ratings = users @ items.T  # matrix multiplication

df = pd.DataFrame(
    predicted_ratings,
    index=["Peter", "Phylla", "Mela"],
    columns=["Pulp Fiction", "Fight Club", "Toy Story", "Notebook"]
)
print("Predicted rating matrix (dot products):")
print(df.round(2))
```

## Finding similar items using item vectors

We can also compare item vectors directly to find similar movies,
without needing any user information.

```python
from sklearn.metrics.pairwise import cosine_similarity

item_names = ["Pulp Fiction", "Fight Club", "Toy Story", "Notebook"]
item_sim = cosine_similarity(items)

sim_df = pd.DataFrame(item_sim, index=item_names, columns=item_names)
print("Item-item similarity (cosine):")
print(sim_df.round(3))
```
